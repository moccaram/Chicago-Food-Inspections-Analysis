```python
# Build a lookup: address -> most common non-null zip seen for that address
zip_lookup = (
    df.dropna(subset=['zip'])
      .groupby('address')['zip']
      .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)

# Fill nulls where a match exists
mask = df['zip'].isna()
df.loc[mask, 'zip'] = df.loc[mask, 'address'].map(zip_lookup)

print(f"Zip nulls remaining after address-match recovery: {df['zip'].isna().sum()}")
```

```python
# Step 1: Split the violations string into individual violation entries
# Isolate, split by pipe, and explode into long format
viol_df = df[['inspection_id', 'violations']].dropna().copy()
viol_df['violations'] = viol_df['violations'].str.split(r'\s*\|\s*')
viol_long = viol_df.explode('violations').copy()
viol_long
```

```python
#Extract code, description, and comments using named capture groups 
pattern = r'^(?P<violation_code>\d+)\.\s+(?P<violation_description>.*?)(?:\s*-\s*Comments:\s*(?P<comments>.*))?$'
extracted = viol_long['violations'].str.extract(pattern)

#Combine the inspection_id with the extracted columns
viol_long = pd.concat([viol_long[['inspection_id']], extracted], axis=1)
viol_long = viol_long.dropna(subset=['violation_code'])

print(f"Step 1: Total individual violations counted: {len(viol_long)}")
```

```python
#STEP 2: Check unique codes vs. unique (code, description) pairs
unique_codes = viol_long['violation_code'].nunique()
unique_pairs = viol_long[['violation_code', 'violation_description']].drop_duplicates().shape[0]


print(f"Step 2: Unique violation codes: {unique_codes}")
print(f"Step 2: Unique (code, description) pairs: {unique_pairs}")
```

```python
# Group descriptions by code and filter for codes with more than 1 variant
drift = viol_long.groupby('violation_code')['violation_description'].unique()
mismatched = drift[drift.apply(len) > 1]

# Display the first few examples to diagnose the issue
for code, descs in mismatched.head(5).items():
    print(f"--- Code {code} ({len(descs)} variants) ---")
    for desc in descs:
        print(f"  > '{desc}'")
    print("\n")
```

```python
# Merge inspection_date from the main df into your long table
viol_dates = viol_long.merge(df[['inspection_id', 'inspection_date']], on='inspection_id', how='left')

# Filter for Code 1 and group by the description to find min/max dates
code_1_check = viol_dates[viol_dates['violation_code'] == '1']
date_ranges = code_1_check.groupby('violation_description')['inspection_date'].agg(['min', 'max'])

print(date_ranges)
```

```python
# Merge inspection_date from the main df into your long table
viol_dates = viol_long.merge(df[['inspection_id', 'inspection_date']], on='inspection_id', how='left')

# Filter for Code 1 and group by the description to find min/max dates
code_1_check = viol_dates[viol_dates['violation_code'] == '10']
date_ranges = code_1_check.groupby('violation_description')['inspection_date'].agg(['min', 'max'])

print(date_ranges)
```

```python
# --- 1. Verify the 2018-07-01 boundary across ALL codes ---
# Group by every unique code + description combo and get their full date ranges
all_ranges = viol_dates.groupby(['violation_code', 'violation_description'])['inspection_date'].agg(['min', 'max']).reset_index()

# A "straddler" is a specific description that was used BOTH before and after the July 1 cutoff
straddlers = all_ranges[(all_ranges['min'] < '2018-07-01') & (all_ranges['max'] >= '2018-07-01')]

print(f"Total (code, description) pairs straddling the July 1, 2018 boundary: {len(straddlers)}")
if len(straddlers) > 0:
    display(straddlers.head())

# --- 2. Add the 'era' column ---
viol_dates['era'] = np.where(viol_dates['inspection_date'] < '2018-07-01', 'pre_2018', 'post_2018')

# Quick check of the split
print("\nViolation counts by era:")
print(viol_dates['era'].value_counts())
```

```python
# --- 1. Final Sanity Check ---
print(f"Total Clean Inspections: {len(df)}")
print(f"Total Parsed Violations: {len(viol_dates)}")

# --- 2. Export ---
# Ensure the processed data directory exists
os.makedirs('data/processed', exist_ok=True)

# Export the clean inspections table (one row per inspection)
df.to_csv('data/processed/inspections_clean.csv', index=False)

# Export the long-format violations table (one row per violation instance)
viol_dates.to_csv('data/processed/violations_long.csv', index=False)

print("\nPhase 1 Complete: Clean datasets exported successfully.")
```
